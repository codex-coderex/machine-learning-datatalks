```python
import pandas as pd
import numpy as np

url = ("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")
df = pd.read_csv(url)
print(df.head())
```

```python
print("Pandas version:", pd.__version__)
print("Number of rows in the dataset:", df.shape[0])
print("Number of unique fuel types:", df['fuel_type'].nunique())
print("Number of columns with missing values:", df.isna().any().sum())
print("Maximum fuel efficiency (MPG) for cars from Asia:", df.loc[df["origin"] == "Asia", 'fuel_efficiency_mpg'].max())

mode_hp = df['horsepower'].mode()[0]

print("Median horsepower:", df['horsepower'].median())  
print("Mode horsepower:", mode_hp)
df["horsepower"] = df["horsepower"].fillna(mode_hp)

print("Median horsepower:", df['horsepower'].median())  
```

```python
vh = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]]
vh_7 = vh.head(7)
X = vh_7.to_numpy()

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print(w.sum())
```
